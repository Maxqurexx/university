### **Список (`list`)**

**Список** — это **упорядоченная изменяемая** коллекция элементов. Элементы могут быть любого типа, и в одном списке можно хранить элементы разных типов.

#### **Создание списка**

#### **Основные методы списка**

- **`append(element)`**: добавляет элемент в конец списка.
- **`insert(index, element)`**: вставляет элемент на заданную позицию.
- **`remove(element)`**: удаляет первый найденный элемент из списка.
- **`pop(index)`**: удаляет элемент по индексу и возвращает его.
- **`sort()`**: сортирует список по возрастанию.
- **`reverse()`**: разворачивает список.
- **`index(element)`**: возвращает индекс первого вхождения элемента.
- **`count(element)`**: возвращает количество вхождений элемента.
- **`extend(iterable)`**: расширяет список, добавляя элементы из другого итерируемого объекта.
- **`clear()`**: очищает список.
- **`copy()`**: создаёт поверхностную копию списка.

#### Пример использования

```python
# Создаем список чисел
numbers = [5, 2, 9, 1]

# Добавляем число в конец списка
numbers.append(7)
print("После добавления 7:", numbers)
```

```python
# Создаем список чисел
numbers = [5, 2, 9, 1]

# Вставляем число на позицию с индексом 2
numbers.insert(2, 6)
print("После вставки 6 на позицию 2:", numbers)
```

```python
# Создаем список чисел
numbers = [5, 2, 9, 1]

# Сортируем список
numbers.sort()
print("После сортировки:", numbers)
```

```python
# Создаем список чисел
numbers = [5, 2, 9, 1]

# Получаем индекс числа 9
index_9 = numbers.index(9)
print("Индекс числа 9:", index_9)
```

#### **Работа со срезами**

```python
# Срез списка: элементы с индекса i (включительно) до j (не включительно)
c = [1, 2, 3, 4, 5, 6, 7, 8, 9]
d = c[1:7:2]  # каждый второй элемент с индекса 1 до 7
print(d)  # [2, 4, 6]

# Переворот списка с помощью среза
reversed_list = c[::-1]
print(reversed_list)  # [9, 8, 7, 6, 5, 4, 3, 2, 1]
```

#### Создание множества

```python
# Множество с элементами
my_set = {1, 2, 3, 4, 5}

# Пустое множество (нельзя использовать {})
empty_set = set()
```

#### Основные методы множества

- **`add(element)`**: добавляет элемент в множество.
- **`remove(element)`**: удаляет элемент; вызывает ошибку, если элемент не найден.
- **`discard(element)`**: удаляет элемент; не вызывает ошибку, если элемент не найден.
- **`pop()`**: удаляет и возвращает случайный элемент.
- **`clear()`**: очищает множество.
- **`union(other_set)`**: объединение множеств.
- **`intersection(other_set)`**: пересечение множеств.
- **`difference(other_set)`**: разность множеств.
- **`symmetric_difference(other_set)`**: симметрическая разность множеств.
- **`copy()`**: создает копию множества.
- **`update(iterable)`**: добавляет несколько элементов в множество.
- **`issubset(other_set)`**: проверяет, является ли множество подмножеством другого.
- **`issuperset(other_set)`**: проверяет, является ли множество надмножеством другого.

#### Пример использования

```python
# Создаем множества
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Добавляем элемент
set_a.add(6)
print("После добавления 6 в set_a:", set_a)
```

```python
# Создаем множества
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Удаляем элемент
set_a.discard(2)
print("После удаления 2 из set_a:", set_a)
```

```python
# Создаем множества
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Объединение множеств
union_set = set_a.union(set_b)
# Альтернативный способ: union_set = set_a | set_b
print("Объединение set_a и set_b:", union_set)
```

```python
# Создаем множества
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Пересечение множеств
intersection_set = set_a.intersection(set_b)
# Альтернативный способ: intersection_set = set_a & set_b
print("Пересечение set_a и set_b:", intersection_set)
```

```python
# Создаем множества
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Разность множеств
difference_set = set_a.difference(set_b)
# Альтернативный способ: difference_set = set_a - set_b
print("Разность set_a и set_b:", difference_set)
```

```python
# Симметрическая разность
symmetric_diff = set_a.symmetric_difference(set_b)
# Альтернативный способ: symmetric_diff = set_a ^ set_b
print("Симметрическая разность:", symmetric_diff)
```

```python
# Проверка на подмножество и надмножество
A = {1, 2, 3, 4, 5}
B = {2, 3, 4}

print("B является подмножеством A:", B.issubset(A))  # или B <= A
print("A является надмножеством B:", A.issuperset(B))  # или A >= B
```

```python
# Удаление дубликатов из списка
list_with_duplicates = [1, 2, 3, 5, 3, 2, 4, 7]
list_unique = list(set(list_with_duplicates))
print("Список без дубликатов:", list_unique)
```

### **Словарь (`dict`)**

**Словарь** — это **неупорядоченная коллекция пар "ключ-значение"**. Ключи должны быть уникальными и неизменяемыми типами данных (например, строки или числа).

#### Создание словаря

```python
# Словарь с данными
person = {
    "имя": "Иван",
    "возраст": 30,
    "город": "Москва"
}

# Пустой словарь
empty_dict = {}
empty_dict = dict()
```

```python
list(person.keys()), list(person.values())
```

#### Основные методы словаря

- **`get(key[, default])`**: возвращает значение по ключу; если ключ не найден, возвращает `default`.
- **`keys()`**: возвращает все ключи словаря.
- **`values()`**: возвращает все значения словаря.
- **`items()`**: возвращает пары "ключ-значение".
- **`update(other_dict)`**: обновляет словарь, добавляя пары из другого словаря.
- **`pop(key[, default])`**: удаляет и возвращает значение по ключу; если ключ не найден, возвращает `default`.
- **`clear()`**: очищает словарь.
- **`copy()`**: создает копию словаря.
- **`setdefault(key[, default])`**: возвращает значение по ключу, если ключ существует; иначе создает ключ со значением `default`.
- **`fromkeys(keys[, value])`**: создает новый словарь с ключами из последовательности и одинаковыми значениями.
- **`popitem()`**: удаляет и возвращает случайную пару (ключ, значение).

#### Пример использования

```python
# Создаем словарь студента
student = {
    "имя": "Мария",
    "факультет": "Математика",
    "курс": 2
}

print(student)
```

```python
# Получаем значение по ключу
faculty = student.get("факультет")
print("Факультет:", faculty)
```

```python
# Добавляем новую пару
student["средний балл"] = 4.5
print("Обновленный словарь:", student)
```

```python
# Получаем все ключи
keys = student.keys()
print("Ключи словаря:", list(keys))
```

```python
# Получаем все значения
values = student.values()
print("Значения словаря:", list(values))
```

```python
# Обновляем словарь
student.update({"курс": 3, "любимый предмет": "Алгебра"})
print("После обновления:", student)
```

```python
# Удаляем элемент
grade = student.pop("средний балл")
print("Удален средний балл:", grade)
print("Словарь после удаления:", student)
```

**Практическое применение**

Словари широко используются для хранения и быстрого доступа к данным по ключу, например, для хранения информации о пользователях, конфигураций или результатов вычислений.
